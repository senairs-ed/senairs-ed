```python
# --- Célula: ROI Calculator ---
def calcular_roi(custo_projeto, beneficio_anual, anos=5):
  # Custo total (incluindo manutenção anual de 20%)
  custo_total = custo_projeto + (custo_projeto * 0.2 * anos)

  # Benefício total
  beneficio_total = beneficio_anual * anos

  # ROI
  roi = ((beneficio_total - custo_total) / custo_total) * 100

  # Payback Period (em meses)
  payback_meses = (custo_projeto / beneficio_anual) * 12

  # NPV (Net Present Value) - taxa de desconto 10%
  taxa_desconto = 0.10
  npv = sum([beneficio_anual / ((1 + taxa_desconto) ** ano) for ano in range(1, anos+1)]) - custo_total

  return {"ROI (%)": round(roi, 2),
          "Payback (meses)": round(payback_meses, 1),
          "NPV (USD)": round(npv, 2),
          "Viável?": "SIM" if npv > 0 else "NÃO"}

# Exemplo: Projeto de Churn Prediction
resultado = calcular_roi( custo_projeto=300000, # $300k
                          beneficio_anual=1200000, # $1.2M/ano (redução de churn)
                          anos=3)

print("--- Análise de Viabilidade ---")
for k, v in resultado.items():
  print(f"{k}: {v}")
```

# Calculadora de ROI e Viabilidade de Projetos de IA

Esta ferramenta calcula a viabilidade financeira de projetos (como modelos de Churn Prediction ou Fraude) utilizando as seguintes métricas:

*   **ROI (Return on Investment):** O percentual de retorno sobre o investimento total (incluindo custos de manutenção).
*   **Payback Period:** O tempo necessário (em meses) para que os benefícios anuais cubram o investimento inicial.
*   **NPV (Net Present Value):** O valor presente dos fluxos de caixa futuros, descontados a uma taxa de 10%, subtraindo o investimento total. Se for positivo, o projeto é considerado financeiramente viável.
