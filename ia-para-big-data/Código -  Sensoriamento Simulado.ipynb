```python
import random, time
import pandas as pd

def coletar_dados(sensores=1000, leituras_por_sensor=100):
    dados = []
    for s in range(sensores):
        for l in range(leituras_por_sensor):
            leitura = {
                'sensor_id': s,
                'temperatura': round(random.uniform(20.0, 80.0), 2),
                'vibracao': round(random.uniform(0.1, 5.0), 2),
                'corrente': round(random.uniform(0.5, 10.0), 2),
                'timestamp': time.time()
            }
            dados.append(leitura)
    return pd.DataFrame(dados)

df = coletar_dados()
df.head()
```

Esse código simula uma operação industrial de coleta massiva.

### 4.8 Extração de features com pandas (janelas de 1 min)

```python
import pandas as pd

df['timestamp'] = pd.to_datetime(df['timestamp'], unit='s')
df = df.set_index('timestamp')

# agregação por janelas móveis
features = df.groupby('sensor_id').resample('1T').agg({
    'temperatura': ['mean', 'std', 'max'],
    'vibracao': ['mean', 'max']
}).reset_index()

features.head()
```
