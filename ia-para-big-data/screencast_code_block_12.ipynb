```python
from evidently.report import Report
from evidently.metric_preset import DataDriftPreset
import pandas as pd
import numpy as np

print("Sucesso! Módulos carregados.")
```

```python
# --- Célula Extra: Drift Detection ---
# Dados de Referência (Treino)
reference_data = pd.DataFrame({
    'feature1': np.random.normal(50, 10, 1000),
    'feature2': np.random.normal(30, 5, 1000),
    'target': np.random.randint(0, 2, 1000)
})

# Dados de Produção (Com Drift!)
production_data = pd.DataFrame({
    'feature1': np.random.normal(70, 10, 1000), # Média mudou!
    'feature2': np.random.normal(30, 5, 1000),
    'target': np.random.randint(0, 2, 1000)
})

# Gerar Relatório de Drift
report = Report(metrics=[DataDriftPreset()])
report.run(reference_data=reference_data, current_data=production_data)

# Salvar HTML
report.save_html("drift_report.html")

print("Relatório de Drift gerado! Abra 'drift_report.html' para visualizar.")
# (Análise) O relatório mostra que feature1 sofreu drift significativo (p-value < 0.05). Isso significa que o modelo precisa ser retreinado com dados novos.
```

## O que é Data Drift?

O **Data Drift** (desvio de dados) ocorre quando as propriedades estatísticas dos dados de entrada mudam ao longo do tempo. Isso é um problema crítico porque modelos de Machine Learning treinados em dados antigos podem perder a precisão ao lidar com novos padrões.

### Como este laboratório funciona:
1. **Dados de Referência**: Representam os dados usados durante o treinamento do modelo.
2. **Dados de Produção**: Representam os dados que o modelo está recebendo agora no mundo real.
3. **Evidently AI**: Utilizamos esta biblioteca para comparar as distribuições de `feature1` e `feature2` entre os dois conjuntos.

### Análise do Relatório:
- Se o **p-value** for menor que 0.05, o Evidently detecta um desvio significativo.
- No nosso exemplo, provocamos um drift proposital na `feature1` (mudando a média de 50 para 70), o que deve ser sinalizado no relatório HTML gerado abaixo.
