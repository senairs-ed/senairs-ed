```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros com incerteza
custo_projeto = np.random.normal(500000, 50000, 10000) # Média $500k, desvio $50k
beneficio_anual = np.random.normal(1200000, 200000, 10000) # Média $1.2M, desvio $200k

# Calcular ROI para cada simulação
roi_simulado = ((beneficio_anual * 3) - custo_projeto) / custo_projeto * 100

# Análise
print(f"ROI Médio: {np.mean(roi_simulado):.2f}%")
print(f"ROI no pior caso (5º percentil): {np.percentile(roi_simulado, 5):.2f}%")
print(f"Probabilidade de ROI > 100%: {(roi_simulado > 100).sum() / len(roi_simulado) * 100:.1f}%")

# Visualizar
plt.hist(roi_simulado, bins=50, edgecolor='black')
plt.title("Distribuição de ROI (Monte Carlo)")
plt.xlabel("ROI (%)")
plt.ylabel("Frequência")
plt.show()
```

### Explicação dos Resultados

A simulação de Monte Carlo realizada acima modela a incerteza nos custos e benefícios para projetar o ROI (Retorno sobre Investimento).

1.  **ROI Médio**: Representa o retorno esperado mais provável considerando todas as variações simuladas.
2.  **Pior Caso (5º Percentil)**: Indica que há apenas 5% de chance de o ROI ser menor que este valor, servindo como uma medida de segurança para o projeto.
3.  **Probabilidade de ROI > 100%**: Mostra a confiança estatística de que o projeto irá, no mínimo, dobrar o capital investido.

O histograma visualiza a densidade dessas possibilidades, permitindo identificar a dispersão e o risco do investimento.
