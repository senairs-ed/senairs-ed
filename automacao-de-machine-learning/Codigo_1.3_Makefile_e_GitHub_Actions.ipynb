# Código 1.3 — Automação com Makefile e GitHub Actions (CI/CD)

**Objetivo:** Demonstrar a padronização de comandos locais com `Makefile` e a criação de uma esteira automatizada de integração contínua (CI) com `GitHub Actions`.

### Passo 1: Padronização de Comandos com Makefile

O `Makefile` abstrai comandos repetitivos do terminal (instalação, lint, testes e execução de treinamento) em alvos curtos.

```python
makefile_content = """
.PHONY: install lint test train run

install:
	pip install --upgrade pip && pip install -r requirements.txt

lint:
	ruff check .

test:
	pytest tests/ -v

train:
	python src/train.py

run:
	uvicorn api:app --reload --port 8000
"""
print("Conteúdo do Makefile:")
print(makefile_content)
```

**Entendendo os alvos e comandos do Makefile:**
- `.PHONY`: Declara alvos que não correspondem a arquivos físicos, garantindo que o Make os execute sempre.
- `install`: Atualiza o gerenciador `pip` e instala os pacotes listados em `requirements.txt`.
- `lint`: Executa o analisador estático de código `ruff` para verificar regras de estilo e boas práticas em Python.
- `test`: Executa a suíte de testes automatizados usando `pytest` na pasta `tests/` com saída detalhada (`-v`).
- `train`: Roda o script principal de treinamento do modelo (`train.py`).
- `run`: Sobe a API FastAPI localmente habilitando o recarregamento automático (`--reload`).

### Passo Passo 2: Configuração da Esteira no GitHub Actions (`.github/workflows/ci.yml`)

O workflow executa automaticamente verificações a cada `push` ou `pull_request` na branch `main`.

```python
workflow_content = """
name: CI Pipeline

on:
  push:
    branches: [ "main" ]
  pull_request:
    branches: [ "main" ]

jobs:
  test:
    runs-on: ubuntu-latest
    steps:
    - uses: actions/checkout@v3

    - name: Configurar Python
      uses: actions/setup-python@v4
      with:
        python-version: "3.11"

    - name: Instalar dependências
      run: make install

    - name: Executar Linter
      run: make lint

    - name: Executar Testes Unitários
      run: make test
"""
print("Conteúdo do Workflow CI:")
print(workflow_content)
```

**Entendendo as instruções do GitHub Actions Workflow:**
- `on: push / pull_request`: Define os gatilhos de disparo automatizado da esteira ao alterar a branch `main`.
- `jobs / runs-on: ubuntu-latest`: Aloca uma máquina virtual Linux Ubuntu temporária nos servidores do GitHub.
- `actions/checkout@v3`: Baixa o código do repositório para dentro do ambiente de execução do runner.
- `actions/setup-python@v4`: Prepara o ambiente Python especificando a versão `3.11`.
- `run: make install / lint / test`: Executa sequencialmente os comandos definidos no `Makefile`.
