**Objetivo:** Avaliar o impacto financeiro de um modelo de Machine Learning comparando custos operacionais, falhas não detectadas e custos de manutenção.

### Passo 1: Função de Cálculo Financeiro

Definimos a função que recebe a matriz de confusão e os valores de custo/benefício do negócio.

```python
import matplotlib.pyplot as plt
import pandas as pd

def calcular_roi_manutencao(
    tp: int,
    fp: int,
    fn: int,
    tn: int,
    custo_preventiva: float = 1500.0,
    custo_falha_catastrofica: float = 12000.0,
) -> dict[str, float]:
    """Calcula o custo total e a economia gerada pelo modelo preditivo."""
    
    # Custo de intervir preventivamente (TP + FP)
    custo_intervencoes = (tp + fp) * custo_preventiva
    
    # Custo severo de deixar uma falha ocorrer sem intervenção (FN)
    custo_falhas = fn * custo_falha_catastrofica
    
    # Custo total operando com o modelo
    custo_total_modelo = custo_intervencoes + custo_falhas
    
    # Custo base sem modelo (todas as falhas reais ocorreriam catastroficamente: TP + FN)
    custo_sem_modelo = (tp + fn) * custo_falha_catastrofica
    
    # Economia líquida obtida
    economia_liquida = custo_sem_modelo - custo_total_modelo
    roi_percentual = (economia_liquida / custo_total_modelo) * 100 if custo_total_modelo > 0 else 0.0

    return {
        "custo_total_modelo": custo_total_modelo,
        "custo_sem_modelo": custo_sem_modelo,
        "economia_liquida": economia_liquida,
        "roi_percentual": roi_percentual,
    }
```

**Entendendo os parâmetros da função financeira:**
- `tp` (Verdadeiro Positivo): Falhas reais detectadas pelo modelo que permitiram manutenção preventiva agendada.
- `fp` (Falso Positivo): Alarmes falsos que geraram inspeções desnecessárias.
- `fn` (Falso Negativo): Falhas reais não detectadas que resultaram em quebra catastrófica da máquina.
- `tn` (Verdadeiro Negativo): Equipamentos operando normalmente e corretamente identificados.
- `custo_preventiva`: Valor financeiro gasto para realizar uma manutenção agendada antecipadamente (ex: R$ 1.500,00).
- `custo_falha_catastrofica`: Prejuízo financeiro de uma parada não planejada com dano ao equipamento (ex: R$ 12.000,00).

### Passo 2: Simulação Operacional e Comparativo Visual de Custos

Executamos a simulação financeira e geramos um gráfico comparativo dos custos.

```python
# Matriz de confusão simulada
tp, fp, fn, tn = 180, 30, 20, 770

resultado = calcular_roi_manutencao(tp, fp, fn, tn)

print("=== RELATÓRIO FINANCEIRO DE ML ===")
print(f"Custo Sem Modelo (Reativo) : R$ {resultado['custo_sem_modelo']:,.2f}")
print(f"Custo Com Modelo (Preditivo): R$ {resultado['custo_total_modelo']:,.2f}")
print(f"Economia Líquida Gerada     : R$ {resultado['economia_liquida']:,.2f}")
print(f"ROI Estimado                : {resultado['roi_percentual']:.2f}%")

# Gráfico Comparativo de Custos
categorias = ["Sem Modelo (Reativo)", "Com Modelo (Preditivo)"]
valores = [resultado["custo_sem_modelo"], resultado["custo_total_modelo"]]

plt.figure(figsize=(7, 4))
barras = plt.bar(categorias, valores, color=["#d9534f", "#5cb85c"])
plt.ylabel("Custo Total (R$)")
plt.title("Comparativo de Custos Operacionais")

for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, yval + 50000, f"R$ {yval:,.2f}", ha="center", va="bottom", fontweight="bold")

plt.tight_layout()
plt.show()
```
