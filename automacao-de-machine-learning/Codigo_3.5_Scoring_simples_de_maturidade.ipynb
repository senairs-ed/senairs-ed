# Código 3.5 — Avaliação do Nível de Maturidade em MLOps

**Objetivo:** Executar um questionário programático para classificar a maturidade da organização em MLOps (Nível 0: Manual, Nível 1: Automação de Pipeline, Nível 2: CI/CD de ML).

### Passo 1: Estrutura do Diagnóstico e Perguntas Chave

Definimos os critérios de pontuação técnica por dimensão MLOps.

```python
critérios_mlops = {
    "Rastreabilidade de Experimentos (MLflow/W&B)": True,
    "Pipelines de Treinamento Automatizados": True,
    "Validação Automatizada de Modelos (Model Gate)": True,
    "Monitoramento contínuo de Data Drift / Concept Drift": False,
    "Retreinamento Automatizado acionado por Triggers": False,
    "CI/CD de Infraestrutura e Código de ML": True,
}

print("Critérios Avaliados:")
for k, v in critérios_mlops.items():
    status = "[X] Sim" if v else "[ ] Não"
    print(f"  {status} - {k}")
```

### Passo 2: Algoritmo de Cálculo e Classificação de Maturidade

Calculamos a pontuação percentual e classificamos o nível da equipe.

```python
# Calcular pontuação
pontos = sum(critérios_mlops.values())
total_possivel = len(critérios_mlops)
score_percentual = (pontos / total_possivel) * 100

# Determinar Nível de Maturidade
if score_percentual < 35:
    nivel = "Nível 0: Processos Manuais e Ad-hoc"
elif score_percentual < 75:
    nivel = "Nível 1: Automação de Pipelines de ML (Pipeline CI/CD Básico)"
else:
    nivel = "Nível 2: MLOps Avançado com CI/CD Total e Monitoramento Ativo"

print("\n=== RESULTADO DO SCORED DE MATURIDADE MLOPS ===")
print(f"Pontuação: {pontos}/{total_possivel} ({score_percentual:.1f}%)")
print(f"Classificação: {nivel}")
```

**Entendendo as variáveis de cálculo de maturidade:**
- `critérios_mlops`: Dicionário chave-valor mapeando as 6 capacidades fundamentais em MLOps com booleanos (`True`/`False`).
- `score_percentual`: Percentual de aderência às práticas de engenharia de ML.
- `nivel`: Classificação baseada no padrão oficial de maturidade MLOps do Google (Nível 0, Nível 1 ou Nível 2).
