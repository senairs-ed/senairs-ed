## 2.3 — Células de infraestrutura


````python
import urllib.request

def criar_imagem_sintetica(nome, destino):
    """
    Cria uma imagem simples caso o download não funcione.
    Essa função é apenas uma segurança para a prática.
    """

    if nome == "smarties":
        # Cria uma imagem branca.
        img = np.full((500, 700, 3), 255, dtype=np.uint8)

        # Cada item contém: posição e cor BGR.
        objetos = [
            ((100, 120), (0, 0, 230)),
            ((230, 130), (0, 200, 0)),
            ((370, 110), (230, 80, 20)),
            ((540, 130), (0, 0, 230)),
            ((160, 280), (230, 80, 20)),
            ((320, 280), (0, 0, 230)),
            ((480, 280), (0, 200, 0)),
            ((610, 300), (230, 80, 20)),
            ((250, 410), (80, 30, 110)),
            ((470, 410), (0, 0, 230)),
        ]

        # Desenha vários círculos.
        for centro, cor in objetos:
            cv2.circle(
                img,
                centro,
                45,
                cor,
                -1
            )

    elif nome == "sudoku":
        # Cria uma imagem com gradiente de iluminação.
        h = 600
        w = 600

        img = np.tile(
            np.linspace(220, 100, w, dtype=np.uint8),
            (h, 1)
        )

        # Desenha uma grade.
        for i in range(10):
            p = int(i * w / 9)

            if i % 3 == 0:
                espessura = 4
            else:
                espessura = 1

            cv2.line(img, (p, 0), (p, h - 1), 40, espessura)
            cv2.line(img, (0, p), (w - 1, p), 40, espessura)

    else:
        # Building - Cria uma cena simples com muitas linhas.
        img = np.full((480, 720, 3), 235, dtype=np.uint8)

        cv2.rectangle(
            img,
            (120, 90),
            (650, 420),
            (180, 180, 180),
            -1
        )

        for y in range(120, 390, 70):
            for x in range(160, 610, 90):
                cv2.rectangle(
                    img,
                    (x, y),
                    (x + 45, y + 40),
                    (50, 50, 50),
                    3
                )

    # Salva a imagem criada.
    cv2.imwrite(str(destino), img)
````

````python
def baixar_ou_criar(nome, url, extensao):
    """
    Tenta baixar uma imagem.
    Se falhar, cria uma imagem sintética.
    """

    # Define onde o arquivo será salvo.
    destino = PASTA / f"{nome}{extensao}"

    try:
        # Só baixa se o arquivo ainda não existir.
        if not destino.exists():
            urllib.request.urlretrieve(url, destino)

        print(nome, "-> download realizado")

    except Exception:
        # Se ocorrer erro de rede, cria uma alternativa sintética.
        criar_imagem_sintetica(nome, destino)

        print(nome, "-> imagem sintética criada")

    # Devolve o caminho do arquivo.
    return destino
````

````python
arquivo_smarties = baixar_ou_criar(
    "smarties",
    URL_SMARTIES,
    ".png"
)

arquivo_sudoku = baixar_ou_criar(
    "sudoku",
    URL_SUDOKU,
    ".png"
)

arquivo_building = baixar_ou_criar(
    "building",
    URL_BUILDING,
    ".jpg"
)
````

## 6.4 — Criar ruído artificial


````python
# Converte a imagem do prédio para cinza.
building_gray = cv2.cvtColor(
    building,
    cv2.COLOR_BGR2GRAY
)
````

````python
# Cria um gerador de números aleatórios.
# A semente 42 permite repetir o mesmo experimento.
rng = np.random.default_rng(42)
````

````python
# Gera uma matriz de ruído com o mesmo tamanho da imagem.
ruido = rng.normal(
    loc=0,              # média do ruído
    scale=25,           # intensidade aproximada
    size=building_gray.shape
)
````

````python
# Converte a imagem para float para somar o ruído.
imagem_float = building_gray.astype(
    np.float32
)

# Soma imagem + ruído.
imagem_com_ruido = imagem_float + ruido
````

````python
# Garante que os valores permaneçam entre 0 e 255.
imagem_com_ruido = np.clip(
    imagem_com_ruido,
    0,
    255
)

# Retorna ao formato uint8.
imagem_com_ruido = imagem_com_ruido.astype(
    np.uint8
)
````

````python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(building_gray, cmap="gray")
plt.title("Imagem original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagem_com_ruido, cmap="gray")
plt.title("Imagem com ruído")
plt.axis("off")

plt.show()
````

# 9 — Contornos e medições


## 9.1 — Encontrar contornos


````python
contornos, _ = cv2.findContours(
    fechamento,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Quantidade bruta de contornos:", len(contornos))
````

## 9.2 — Criar uma imagem para desenhar os resultados


````python
resultado = smarties.copy()
````

## 9.3 — Criar uma lista para guardar informações


````python
objetos = []
````

## 9.4 — Percorrer os contornos


````python
for contorno in contornos:

    # Calcula a área interna do contorno.
    area = cv2.contourArea(
        contorno
    )

    # Se a área for muito pequena,
    # ignoramos esse contorno.
    if area < 300:
        continue

    # Calcula uma caixa retangular ao redor do objeto.
    x, y, largura_obj, altura_obj = cv2.boundingRect(
        contorno
    )

    # Guarda os dados em uma lista.
    objetos.append(
        [
            x,
            y,
            largura_obj,
            altura_obj,
            area
        ]
    )

    # Desenha a caixa verde.
    cv2.rectangle(
        resultado,
        (x, y),
        (x + largura_obj, y + altura_obj),
        (0, 255, 0),
        2
    )
````

## 9.5 — Mostrar os objetos detectados


````python
resultado_rgb = cv2.cvtColor(
    resultado,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(8, 6))
plt.imshow(resultado_rgb)
plt.title("Objetos vermelhos detectados")
plt.axis("off")
plt.show()
````

## 10.3 — criar duas máscaras para o vermelho


````python
mask_pipeline_1 = cv2.inRange(
    entrada_hsv,
    np.array([0, 90, 70]),
    np.array([10, 255, 255])
)
````

````python
mask_pipeline_2 = cv2.inRange(
    entrada_hsv,
    np.array([170, 90, 70]),
    np.array([179, 255, 255])
)
````

## 10.4 — combinar as máscaras


````python
mask_pipeline = cv2.bitwise_or(
    mask_pipeline_1,
    mask_pipeline_2
)
````

## 10.5 — limpar a máscara


````python
mask_pipeline = cv2.morphologyEx(
    mask_pipeline,
    cv2.MORPH_OPEN,
    kernel
)
````

````python
mask_pipeline = cv2.morphologyEx(
    mask_pipeline,
    cv2.MORPH_CLOSE,
    kernel
)
````

## 10.6 — encontrar contornos


````python
contornos_pipeline, _ = cv2.findContours(
    mask_pipeline,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)
````

## 10.7 — preparar a imagem de saída


````python
saida_pipeline = entrada.copy()

quantidade = 0
````

## 10.8 — filtrar e desenhar os objetos


````python
for contorno in contornos_pipeline:

    # Calcula a área.
    area = cv2.contourArea(
        contorno
    )

    # Ignora regiões pequenas.
    if area < 300:
        continue

    # Conta um objeto válido.
    quantidade = quantidade + 1

    # Calcula bounding box.
    x, y, w, h = cv2.boundingRect(
        contorno
    )

    # Desenha a bounding box.
    cv2.rectangle(
        saida_pipeline,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )
````

## 10.9 — visualizar entrada, máscara e resultado


````python
# Conversões para exibição no Matplotlib.
entrada_rgb = cv2.cvtColor(
    entrada,
    cv2.COLOR_BGR2RGB
)

saida_rgb = cv2.cvtColor(
    saida_pipeline,
    cv2.COLOR_BGR2RGB
)
````

````python
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.imshow(entrada_rgb)
plt.title("1. Entrada")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(mask_pipeline, cmap="gray")
plt.title("2. Máscara")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(saida_rgb)
plt.title("3. Resultado")
plt.axis("off")

plt.show()
````

## 11.5 — Upload opcional de uma imagem própria


````python
def carregar_imagem_do_computador():
    """
    Abre a janela de upload do Google Colab
    e devolve a imagem em formato OpenCV.
    """

    # Ferramenta de upload do Colab.
    from google.colab import files

    # Abre a seleção de arquivos.
    enviados = files.upload()

    # Pega o nome do primeiro arquivo enviado.
    nome = next(
        iter(enviados)
    )

    # Converte os bytes para um array NumPy.
    dados = np.frombuffer(
        enviados[nome],
        dtype=np.uint8
    )

    # Decodifica o arquivo como imagem colorida.
    imagem = cv2.imdecode(
        dados,
        cv2.IMREAD_COLOR
    )

    return imagem
````

# Leitura do pipeline em linguagem simples

1. recebemos uma imagem;
2. mudamos sua representação para HSV;
3. selecionamos pixels vermelhos;
4. criamos uma máscara;
5. limpamos a máscara;
6. encontramos regiões;
7. descartamos regiões pequenas;
8. desenhamos caixas;
9. contamos objetos.

Esse é um pipeline clássico de Visão Computacional.

## Ligação com a teoria

Na teoria usamos a ideia:

**aquisição → pré-processamento → segmentação → características → resultado**

Na prática executamos:

**imagem → HSV → máscara → morfologia → contornos → bounding boxes → contagem**
